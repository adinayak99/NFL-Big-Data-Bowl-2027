```python
import pandas as pd
from summarytools import dfSummary, tabset

df_players = pd.read_csv('data/players.csv')
df_physical_stats = pd.read_csv('data/combine_results.csv')
df_sensor_stats = pd.read_csv('data/combine_tracking.csv')
df_player_career_stats = pd.read_csv('data/player_career_successes.csv')
df_player_play = pd.read_csv('data/player_play.csv')
df_games_info = pd.read_csv('data/games.csv')
df_2023_game_tracking = pd.read_csv('data/game_tracking_2023.csv')
df_2024_game_tracking = pd.read_csv('data/game_tracking_2024.csv')
df_2025_game_tracking = pd.read_csv('data/game_tracking_2025.csv')
```

<p>The following files have the following information:</p>
<ol style="line-height: 2">
    <b><li>players.csv:</b> Contains demographic, collegiate, and draft metadata for the 510 rookie prospects.</li>
    <b><li>combine_results.csv:</b> Contains official anthropometric measurements, standardized physical testing splits, and Next Gen Stats draft grades from the NFL Scouting Combine.</li>
    <b><li>combine_tracking.csv:</b> Contains 10 Hz high-frequency optical sensor tracking across 6,310 Combine drill attempts, including 40-yard dash sprints and position-specific agility/route drills.</li>
    <b><li>player_career_successes.csv:</b> Contains cumulative career volume statistics, snap counts, and accolades earned by the 510 prospects during their rookie and subsequent seasons.</li>
    <b><li>player_play.csv:</b> Contains granular in-game Next Gen Stats and play-by-play metrics for every snap involving the rookie cohort across the 2023, 2024, and 2025 seasons.</li>
    <b><li>games.csv:</b> Contains schedule, season, and team matchup metadata for all 1,002 NFL regular-season and playoff games across 2023 to 2025.</li>
    <b><li>Files 7, 8, 9: game_tracking_2023.csv, game_tracking_2024.csv, game_tracking_2025.csv:</b> Contain 10 Hz high-frequency in-game optical tracking frames for all rookie cohort prospects across each respective NFL season.</li>
</ol>

```python
tabset({"Players": dfSummary(df_players).to_html(),
        "Physical Stats": dfSummary(df_physical_stats).to_html(),
        "Sensor Stats": dfSummary(df_sensor_stats).to_html(),
        "Player Career Stats": dfSummary(df_player_career_stats).to_html(),
        "Player Play": dfSummary(df_player_play).to_html(),
        "Games Info": dfSummary(df_games_info).to_html(),
        "2023 Game Tracking": dfSummary(df_2023_game_tracking).to_html(),
        "2024 Game Tracking": dfSummary(df_2024_game_tracking).to_html(),
        "2025 Game Tracking": dfSummary(df_2025_game_tracking).to_html()})
```

<p>We start by diving the drafted players in offensive and defensive players:</p>
<ol>
    <li>Offensive players</li>
    <ul>
        <li>WR (Wide Receiver)</li>
        <li>T (Tackle)</li>
        <li>G (Guard)</li>
        <li>C (Center)</li>
        <li>TE (Tight End)</li>
        <li>RB (Running Back)</li>
        <li>FB (Fullback)</li>
    </ul>
    <li>Defensive players</li>
    <ul>
        <li>CB (Cornerback)</li>
        <li>DT (Defensive Tackle)</li>
        <li>DE (Defensive End)</li>
        <li>OLB (Outside Linebacker)</li>
        <li>SS (Strong Safety)</li>
        <li>FS (Free Safety)</li>
        <li>NT (Nose Tackle)</li>
        <li>DB (Defensive Back - umbrella term for cornerbacks and safeties)</li>
        <li>ILB (Inside Linebacker)</li>
        <li>MLB (Middle Linebacker)</li>
    </ul>

```python
#Splitting the players in offense and defense dataframes
df_off_players = df_players[df_players['nfl_position'].isin(['WR','T', 'G', 'C', 'TE', 'RB', 'FB'])]

df_def_players = df_players[df_players['nfl_position'].isin(['CB', 'DT', 'DE', 'OLB', 'SS', 'FS', 'NT', 'DB', 'ILB', 'MLB'])]

tabset({"Offensive Players": dfSummary(df_off_players).to_html(),
        "Defensive Players": dfSummary(df_def_players).to_html()})
```

<p>With this categorization, we can analyze the strategies and evaluate accordingly:</p>
